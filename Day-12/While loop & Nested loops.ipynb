```python
# Finding Maximum element in list
l=[90,56,3,2,15,789,7]
maximum=float('-inf')
for num in l:
    if num>maximum:
        maximum=num
print(maximum)
```

```python
# Finding that number is prime number or not
n=int(input())
c=0
for i in range(1,n+1):
    if n%i==0:
        c+=1
if c==2:
    print('Prime Number')
else:
    print('Not Prime Number')
```

```python
# Printing counting of Prime numbers in python
l=[2,3,4,5,6,7,8,9]
a=0
for i in l:
    c=0
    for j in range(1,i+1):
        if i%j==0:
            c+=1
    if c==2:
        a+=1
print(a)
```
